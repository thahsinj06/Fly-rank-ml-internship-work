## 3. Quick look at the data (2-3 real numbers)

*The starter dataset contains 30,000 pages, of which 16,262 (54.21%) are currently showing a downward trend. In addition, 9,961 declining pages have at least 500 impressions, indicating that a substantial number of pages have enough visibility for prioritisation to matter. This supports investigating a system that helps an editorial team identify which pages should be reviewed first.*

```python
import pandas as pd

url = "https://raw.githubusercontent.com/thahsinj06/Fly-rank-ml-internship-work/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(url)

total_pages = len(df)

declining_pages = (df["trend_direction"] == "down").sum()
decline_pct = declining_pages / total_pages * 100

high_visibility_decline = (
    (df["impressions_90d"] >= 500) &
    (df["trend_direction"] == "down")
).sum()

print(f"Total pages: {total_pages:,}")
print(f"Pages showing decline: {declining_pages:,} ({decline_pct:.2f}%)")
print(f"High-visibility pages in decline: {high_visibility_decline:,}")
```
